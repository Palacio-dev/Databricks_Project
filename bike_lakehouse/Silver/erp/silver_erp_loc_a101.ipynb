# Initialization

```python
import pyspark.sql.functions as F
from pyspark.sql.types import StringType, DateType
from pyspark.sql.functions import trim, col
from pyspark.sql.window import Window
```

# Reading from Bronze

```python
df = spark.table("workspace.bronze.erp_loc_a101")
```

# Data transformations

```python
df.display()
df.printSchema()
```

Analyzing the table, we can clearly see that some rows have the contrys full name and others the acronym. This way, we should normalize it. And also, we can rename the columns as usual.

```python
df.select("CNTRY").distinct().display()
```

# Trimming

```python
for field in df.schema.fields:
    if isinstance(field.dataType, StringType):
        df = df.withColumn(field.name, trim(col(field.name)))
```

# Normalizing the country

```python
df = df.withColumn(
    "cntry",
    F.when(col("cntry") == "DE", "Germany")
     .when(col("cntry").isin("US", "USA"), "United States")
     .when((col("cntry") == "") | col("cntry").isNull(), "n/a")
     .otherwise(col("cntry"))
)
```

# Renaming Columns

```python
RENAME_MAP = {
    "cid": "customer_number",
    "cntry": "country"
}
for old_name, new_name in RENAME_MAP.items():
    df = df.withColumnRenamed(old_name, new_name)
     
```

# Sanity check of dataframe

```python
df.limit(10).display()
```

# Writing the Silver Table

```python
df.write.mode("overwrite").format("delta").saveAsTable("workspace.silver.erp_customer_location")
```
